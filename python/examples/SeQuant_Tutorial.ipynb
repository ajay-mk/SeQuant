# SeQuant Python Interface Tutorial

This notebook demonstrates the SeQuant Python interface for symbolic tensor algebra in quantum many-body theory.

## Installation

First, make sure SeQuant is built with Python support and the build directory is in your `PYTHONPATH`:

```bash
cmake -B build -S . -DSEQUANT_PYTHON=ON
cmake --build build --target python-sequant
export PYTHONPATH=/path/to/SeQuant/build/python:$PYTHONPATH
```

## Basic Usage

### Importing SeQuant

````python
import sequant as sq
from sequant.mbpt import CC, Ansatz, np, nh

print(f"SeQuant version: {sq.__version__}")
````

### Context Configuration

SeQuant uses a context to manage computational settings like vacuum type and basis.

````python
from sequant.jupyter import show_context

# Show current context
show_context()
````

````python
# Configure for spin-free CCSD
sq.set_vacuum(sq.Vacuum.SingleProduct)  # Fermi vacuum
sq.set_spbasis(sq.SPBasis.Spinfree)     # Spin-free basis

show_context()
````

## Coupled Cluster Theory

### CCSD Equations

Derive CCSD T amplitude equations:

````python
# Create CCSD engine (rank 2)
cc = CC(2)

# Derive T amplitude equations
t_eqs = cc.t()

print(f"Number of equations: {len(t_eqs)}")
print(f"T1 equation has {t_eqs[1].size()} terms")
print(f"T2 equation has {t_eqs[2].size()} terms")
````

Display the equations (they will automatically render as LaTeX):

````python
# T1 equation
t_eqs[1]
````

````python
# T2 equation (may be long!)
t_eqs[2]
````

### Using Display Helpers

For nicer formatting:

````python
from sequant.jupyter import display_eqs

display_eqs({
    "T₁ Amplitude Equation": t_eqs[1],
    "T₂ Amplitude Equation": t_eqs[2]
})
````

### Lambda Equations

Derive Λ amplitude equations for properties:

````python
lambda_eqs = cc.lambda_()

display_eqs({
    "Λ₁ Equation": lambda_eqs[1],
    "Λ₂ Equation": lambda_eqs[2]
})
````

### EOM-CCSD

Equation-of-Motion CCSD for excited states:

````python
# Excitation energies (EE): 2 particles, 2 holes
r_eqs = cc.eom_r(np(2), nh(2))

print(f"Number of R equations: {len(r_eqs)}")
print(f"R0 has {r_eqs[0].size()} terms")
print(f"R1 has {r_eqs[1].size()} terms")
print(f"R2 has {r_eqs[2].size()} terms")
````

````python
# Display R1 equation
r_eqs[1]
````

### Unitary Coupled Cluster

Use unitary ansatz exp(T - T†):

````python
# Unitary CCSD
ucc = CC(2, Ansatz.U)

# Higher commutator rank often needed for UCC
ucc_t = ucc.t(commutator_rank=6)

print(f"UCC T1 has {ucc_t[1].size()} terms")
print(f"UCC T2 has {ucc_t[2].size()} terms")
````

## Expression Manipulation

### Simplification

````python
from sequant import simplify, canonicalize, expand
from sequant.jupyter import compare_eqs

# Get an equation
expr = t_eqs[1]

# Canonicalize it
canonical = canonicalize(expr)

# Simplify it  
simplified = simplify(expr)

compare_eqs(expr, simplified, "Original", "Simplified")
````

### Spin-Tracing

Transform spin-free equations to spin-orbital form:

````python
from sequant.mbpt import closed_shell_CC_spintrace

# Spin-trace T2 equation
t2_spin = closed_shell_CC_spintrace(t_eqs[2])

print(f"Spin-traced T2 has {t2_spin.size()} terms")
t2_spin
````

## CCSDT Example

Triple excitations:

````python
# CCSDT
ccsdt = CC(3)

# This may take a while...
ccsdt_t = ccsdt.t()

print(f"T1 terms: {ccsdt_t[1].size()}")
print(f"T2 terms: {ccsdt_t[2].size()}")
print(f"T3 terms: {ccsdt_t[3].size()}")

display_eqs({
    "CCSDT T₁": ccsdt_t[1],
    "CCSDT T₂": ccsdt_t[2],
    "CCSDT T₃": ccsdt_t[3]
})
````
